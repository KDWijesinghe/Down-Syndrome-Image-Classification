```python
import tensorflow as tf

MODEL_PATH = "Results/pattern_classifier_final.keras"

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully!")
```

```python
#Create the prediction function for Gradio
import numpy as np
from PIL import Image

def gradio_predict(image):
    if image is None:
        return "Please upload an image."

    # Make sure image is RGB
    image = image.convert("RGB")

    # Resize to the size used during training
    image = image.resize((224, 224))

    # Convert to NumPy array
    img_array = np.array(image, dtype=np.float32)

    # Add batch dimension
    img_array = np.expand_dims(img_array, axis=0)

    # Predict
    probability = model.predict(img_array, verbose=0)[0][0]

    if probability >= 0.5:
        prediction = "Down Syndrome"
        confidence = probability
    else:
        prediction = "Healthy"
        confidence = 1 - probability

    return (
        f"Prediction: {prediction}\n"
        f"Confidence: {confidence * 100:.2f}%\n"
        f"Raw score: {probability:.4f}"
    )

print("Prediction function ready!")
```

```python
#Build the drag-and-drop GUI
import gradio as gr

demo = gr.Interface(
    fn=gradio_predict,
    inputs=gr.Image(
        type="pil",
        label="Drop or Upload an Image"
    ),
    outputs=gr.Textbox(
        label="Prediction Result",
        lines=4
    ),
    title="Identification Of Down Syndrome",
    description=(
        "Upload or drag-and-drop an image."
    )
)

print("GUI created successfully!")
```

```python
demo.launch(
    share=True,
    debug=True
)
```
